```python
import os
import subprocess
from pathlib import Path

def parse_all_pdfs():
    """
    自動尋找並使用 mineru 解析所有 PDF 檔案。
    """
    # --- 設定 ---
    input_dir = Path("C:/Users/User/Desktop/Compliance-GraphRAG/UnstructuredData_Transformation_Pipeline/SPM/EU")
    output_dir = Path("C:/Users/User/Desktop/Compliance-GraphRAG/UnstructuredData_Transformation_Pipeline/out_md_mineru")

    # 確保輸出目錄存在
    output_dir.mkdir(exist_ok=True)

    print("🔍 開始解析 PDF 檔案...")
    print(f"   來源: {input_dir}")
    print(f"   輸出: {output_dir}")
    print("-" * 50)

    # 遞迴尋找所有 .pdf 檔案
    pdf_files = list(input_dir.glob("**/*.pdf"))
    
    if not pdf_files:
        print("⚠️ 在來源目錄中找不到任何 PDF 檔案。")
        return

    for i, pdf_path in enumerate(pdf_files):
        print(f"📄 ({i+1}/{len(pdf_files)}) 正在處理: {pdf_path.name}")
        
        # 建立並執行 mineru 命令
        command = [
            "mineru",
            "-p", str(pdf_path),
            "-o", str(output_dir)
        ]
        
        try:
            result = subprocess.run(command, check=True, capture_output=True, text=True, encoding='utf-8')
            print(result.stdout)
            print("✅ 完成.")
        except FileNotFoundError:
            print("❌ 錯誤: 'mineru' 命令未找到。請確認已安裝並在系統 PATH 中。")
            return
        except subprocess.CalledProcessError as e:
            print(f"❌ 處理失敗: {pdf_path.name}")
            print(f"   錯誤訊息: {e.stderr}")
        
        print("")

    print("🎉 所有 PDF 解析完成！")

parse_all_pdfs()
```
